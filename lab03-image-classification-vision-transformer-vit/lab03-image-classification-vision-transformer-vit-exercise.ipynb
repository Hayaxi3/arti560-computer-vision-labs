```python
# Import Libraries
import requests
import numpy as np
from PIL import Image
import keras
import keras_hub
```

```python
# Load ViTImageClassifierPreprocessor (vit_base_patch16_224_imagenet preset)
preprocessor = keras_hub.models.ViTImageClassifierPreprocessor.from_preset(
    "vit_base_patch16_224_imagenet"
)

# Load ViTImageClassifier (vit_base_patch16_224_imagenet preset)
model = keras_hub.models.ViTImageClassifier.from_preset (
    "vit_base_patch16_224_imagenet",
    # Preprocess the images
    preprocessor=preprocessor
)
```

```python
# Define the images
TEST_IMAGES = [
    {"path": "/content/images/Golden_Retriever.jfif", "true_class": "Golden retriever"},
    {"path": "/content/images/tiger_cat.jfif", "true_class": "Tiger cat"},
    {"path": "/content/images/tiger_shark.jpg", "true_class": "Tiger shark"},
    {"path": "/content/images/strawberry.jfif", "true_class": "Strawberry"},
    {"path": "/content/images/Soccer_Ball.jfif", "true_class": "Soccer ball"},
]
```

```python
# Load ImageNet classes name
classes_url = "https://raw.githubusercontent.com/pytorch/hub/master/imagenet_classes.txt"
classes = requests.get(classes_url).text.splitlines()
```

```python
# Predict classes and display results
for i, item in enumerate(TEST_IMAGES):
    print(f"\n-- Image {i+1}: {item['true_class']} --")

    # Load the images 
    img = Image.open(item['path']).convert('RGB')
    img_array = np.array(img, dtype=np.uint8)        # (H, W, 3) 
    img_batch = np.expand_dims(img_array, axis=0)        # (1, H, W, 3)

    # Prediction
    logits = model.predict(img_batch, verbose=0)      
    probs = keras.ops.softmax(logits, axis=-1)
    probs_np  = np.array(probs)[0]                      
    predicted_id  = np.argmax(probs_np)
    predicted_label = classes[predicted_id]
    confidence = probs_np[predicted_id]

    # Display true and predicted labels and confidence 
    print(f'  True class : {item["true_class"]}')
    print(f'  Predicted  : {predicted_label}  (conf: {confidence:.2%})')
```
